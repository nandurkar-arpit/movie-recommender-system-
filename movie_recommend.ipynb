```python
import nltk 
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer 
from nltk.tokenize import word_tokenize
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
# Text Preprocessing
def preprocessing_txt(text):
    # Lower Casing
    text = str(text).lower()

    # Remove Punctuations
    text = re.sub(r'[^\w\s]', '', text)

    # Tokenize the word
    words  = word_tokenize(text)

    # remove stopwords
    words = [word for word in words if word not in stop_words]

    # limitizing the words
    words = [lemmatizer.lemmatize(word) for word  in words]

    return " ".join(words)
```

```python
df['tags'] = df['tags'].apply(preprocessing_txt)
```

```python
df['tags'][1]
```

```python
# reseting all the index
df = df.reset_index(drop = True)
```

```python
# get indices for specific feature 
indicies = pd.Series(df.index, index = df['title']).drop_duplicates()
```

```python
indicies
```

```python
# Vectorization
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(stop_words = 'english', max_features = 50000, ngram_range = (1,2))
```

```python
tfidf_matrix = tfidf.fit_transform(df['tags'])
```

```python
# Cosine Similarity
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title, n = 10):    # n = numbers of movies to be recommended     
    if title not in indicies:
        return ['Movie  not Found']

    idx = indicies[title]
    sim_score = cosine_similarity(tfidf_matrix[idx], tfidf_matrix).flatten()
    similar_idx = sim_score.argsort()[::-1][1:n+1]   # give the top 10 similarity 
    return df[['title']].iloc[similar_idx]
```

```python
recommend('Toy Story')
```
